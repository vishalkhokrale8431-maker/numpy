```python
#time of exicuting between pythonlist and np.arrray
```

```python
#list
a=[i for i in range(10000000)]
b=[i for i in range(10000000)]
c=[]
import time 
start=time.time()
for i in range(len(a)):
    c.append(a[i]+b[i])
print(time.time()-start)
```

```python
#np.array
import numpy as np
a=np.arange(10000000)
b=np.arange(10000000)
start=time.time()
c=a+b
print(time.time()-start)
```
